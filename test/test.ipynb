```python
def gen_cos(a1:int,a2:int,a3:int):
    cos1 = (a2**2+a3**2-a1**2)/(2*a2*a3)
    cos2 = (a3**2+a1**2-a2**2)/(2*a3*a1)
    cos3 = (a1**2+a2**2-a3**2)/(2*a1*a2)
    return cos1, cos2, cos3
print(gen_cos(5,8,7))
print(gen_cos(3,8,7))
print(gen_cos(3,5,7))
print(gen_cos(6,10,14))
```

```python
import math

triples = []
# suppose that a1 <= a2 <= a3
for a1 in range(1, 20):
    for a2 in range(a1, 20):
        for a3 in range(a2, 20):
            gcd = math.gcd(math.gcd(a1,a2),a3)
            b1, b2, b3 = a1//gcd, a2//gcd, a3//gcd
            if b1==b2==b3:
                continue
            if (b1,b2,b3) in triples:
                continue
            is_acute = b3**2 < b1**2 + b2**2
            ttype = 'acute' if is_acute else 'obtuse'
            cos1, cos2, cos3 = gen_cos(b1, b2, b3)
            if abs(cos1 - 0.5) < 0.0001 or abs(cos2 - 0.5) < 0.0001 or abs(cos3 - 0.5) < 0.0001:
                triples.append((b1,b2,b3))
                print(f"60도: ({b1}, {b2}, {b3}), {ttype}")
            if abs(cos1 + 0.5) < 0.0001 or abs(cos2 + 0.5) < 0.0001 or abs(cos3 + 0.5) < 0.0001:
                triples.append((b1,b2,b3))
                print(f"120도: ({b1}, {b2}, {b3}), {ttype}")
```

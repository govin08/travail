# 3. 벤 다이어그램

```python
from matplotlib_venn import venn2
SA = {'a', 'b', 'c'}
SD = {'a', 'b', 'd', 'e'}

v = venn2([SA, SD], ('Set A', 'Set D'))
```

```python
from matplotlib_venn import venn2
SA = {'a', 'b', 'c'}
SD = {'a', 'b', 'd', 'e'}

only_A = SA - SD           # A에만 있는 원소
only_D = SD - SA           # D에만 있는 원소  
both = SA & SD             # 교집합 원소

v = venn2([SA, SD], ('Set A', 'Set D'))

# 자동으로 원소 표시
v.get_label_by_id('10').set_text(', '.join(sorted(only_A)))
v.get_label_by_id('01').set_text(', '.join(sorted(only_D)))
v.get_label_by_id('11').set_text(', '.join(sorted(both)))
```

```python
import pyautogui
import time
x1, y1 = 1315, 756
x2, y2 = -517, 884
button_positions = {1:(x1,y1), 2:(x2,y2)}

def send_star(monitor_idx:int, num:str):
    x_star, y_star = button_positions[monitor_idx]
    pyautogui.click(x_star, y_star)
    time.sleep(0.1)

    pyautogui.press('backspace', presses=2, interval=0.05)
    pyautogui.write(num, interval=0.05)
    pyautogui.press('tab', presses=4, interval=0.05)

    pyautogui.write("잘 들었습니다.", interval=0.05)

    pyautogui.press('tab', presses=3, interval=0.05)
    pyautogui.press('enter')

```

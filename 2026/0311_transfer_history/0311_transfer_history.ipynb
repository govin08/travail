```python
import pandas as pd
```

```python
transactions_1 = pd.read_csv('transactions_1.csv', index_col=False)
transactions_2 = pd.read_csv('transactions_2.csv', index_col=False)
transactions_3 = pd.read_csv('transactions_3.csv', index_col=False)
```

```python
transactions_1
```

```python
df1 = pd.concat([transactions_1, transactions_2])
df2 = transactions_3.copy()
df1['kind'] = 1
df2['kind'] = 2
display(df1.head())
display(df2.head())
```

```python
df = pd.concat([df1, df2], ignore_index=True)
df.replace({'이회숙':'이희숙', '영마 상환':'엄마 상환'}, inplace=True)
duplicated_mask = df.duplicated(subset=['거래일자', '거래시각'], keep=False)
df.loc[duplicated_mask, 'kind'] = 3
df = df.sort_values(['거래일자', '거래시각']).reset_index(drop=True).drop_duplicates(['거래일자', '거래시각'])
df
```

```python
df['거래금액'].value_counts()
```

```python
df = df[df['거래금액'].isin([500000,500300,400000,400300])]
# df.replace({400300:400000,500300:500000},inplace=True)
# df
```

```python
print(df['거래금액'].value_counts())
print(df['거래금액'].count())
print(df['거래금액'].sum())
```

```python
df = df[['거래일자', '거래시각', '거래종류', '입지내역', '적요', '통화', '거래금액', '단말','거래점','상대계좌']]
df.to_csv('transfer_history.csv',index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.lib import colors
from reportlab.platypus import Table, TableStyle

# 한글 폰트 설정 (필요 시)
# pdfmetrics.registerFont(TTFont('NanumGothic', 'NanumGothic.ttf'))

# DataFrame을 PDF로
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle, Paragraph
from reportlab.lib.pagesizes import letter


pdf = SimpleDocTemplate(df, pagesize=A4)
elements = []

# 데이터 준비
data = [df.columns.to_list()] + df.values.tolist()

# 테이블 생성
table = Table(data)
table.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
    ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
    ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
    ('FONTSIZE', (0, 0), (-1, 0), 10),
    ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
    ('GRID', (0, 0), (-1, -1), 1, colors.black)
]))

elements.append(table)
pdf.build(elements)
```

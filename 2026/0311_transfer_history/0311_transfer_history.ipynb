```python
import pandas as pd
```

```python
transactions_1 = pd.read_csv('transactions_1.csv', index_col=False)
transactions_2 = pd.read_csv('transactions_2.csv', index_col=False)
transactions_3 = pd.read_csv('transactions_3.csv', index_col=False)
```

```python
transactions_1
```

```python
df1 = pd.concat([transactions_1, transactions_2])
df2 = transactions_3.copy()
df1['kind'] = 1
df2['kind'] = 2
display(df1.head())
display(df2.head())
```

```python
df = pd.concat([df1, df2], ignore_index=True)
df.replace({'이회숙':'이희숙', '영마 상환':'엄마 상환'}, inplace=True)
duplicated_mask = df.duplicated(subset=['거래일자', '거래시각'], keep=False)
df.loc[duplicated_mask, 'kind'] = 3
df = df.sort_values(['거래일자', '거래시각']).reset_index(drop=True).drop_duplicates(['거래일자', '거래시각'])
df
```

```python
df_duplicated = df.copy()
```

```python
df = df_duplicated[['거래일자', '적요', '거래금액', '상대계좌']]
df = df[df['거래금액'].isin([400000,400300,500000,500300])]
df['적요'] = df['적요'].replace('이회숙', '이희숙')
new_rows = [
    {
        '거래일자': '2024-04-02',
        '적요': '엄마 상환',
        '거래금액': 1000300,
        '상대계좌': '71404250936 이희숙',
    },
    {
        '거래일자': '2024-05-04',
        '적요': '엄마 상환',
        '거래금액': 1000300,
        '상대계좌': '71404250936 이희숙',
    },
    {
        '거래일자': '2024-05-04',
        '적요': '엄마 상환',
        '거래금액': 5000000,
        '상대계좌': '71404250936 이희숙',
    },
    {
        '거래일자': '2024-06-01',
        '적요': '엄마 상환',
        '거래금액': 1000000,
        '상대계좌': '71404250936 이희숙',
    }    
]
df = pd.concat([df, pd.DataFrame(new_rows)], ignore_index=True)
df['거래일자'] = pd.to_datetime(df['거래일자'], format='mixed').dt.strftime('%Y-%m-%d')
df['거래금액_rounded'] = df['거래금액'].round(-3)
df['거래금액_cumsum'] = df['거래금액_rounded'].cumsum()
df
```

```python
display(df['거래금액_rounded'].value_counts())
print(df['거래금액_rounded'].sum())
```

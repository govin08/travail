```python
# sec 25
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'])
users = df['user'].unique()
user2avgt = {}
for user in users:
    dfu = df[df.user==user].sort_values(by=['주문시간'])
    time_max = dfu['주문시간'].max()
    time_min = dfu['주문시간'].min()
    diff = (time_max - time_min).days
    if diff != 0:
        user2avgt[user] = diff
avgt = np.mean(list(user2avgt.values()))
users_over_avg = [user for user in user2avgt.keys() if user2avgt[user] > avgt]
print(len(users_over_avg))
```

```python
# sec 30
# 1. 주말 주문 건수와 평일 주문 건수를 구하시오
# 2. 주말 주문 건수와 평일 주문 건수의 차이를 절대값으로 구하고 정수형으로 구하시오.
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'])
def is_weekend(DT):
    dow = DT.day_of_week
    if dow in [5,6]:
        return True
    elif dow in [0,1,2,3,4]:
        return False
df['주말'] = df['주문시간'].apply(is_weekend)
num_weekend = df['주말'].sum()
num_weekday = (~df['주말']).sum()
print(int(abs(num_weekend - num_weekday)))
```

```python
# sec 36
# 1. 결측된 판매금액은 해당 지역의 평균 판매금액으로 결측값을 대체하시오.
# 2. 각 거래마다 판매금액과 해당 지역의 평균 판매금액의 차이를 절댓값으로 구하시오.
# 3. 각 지역에서 차이값의 평균을 구한 후, 이 값이 가장 큰 지역의 지역코드를 구하시오.
df = pd.read_csv('sales.csv')
code2avg = dict(df.groupby('지역코드')['판매금액'].mean())
# print(code2avg)
for i, row in df.iterrows():
    sale = row['판매금액']
    code = row['지역코드']
    if pd.isna(sale):
        df.loc[i, '판매금액'] = code2avg[code]
df['지역평균'] = df['지역코드'].map(code2avg)
df['차이'] = abs(df['지역평균'] - df['판매금액'])
dfg = df.groupby(by='지역코드')['차이'].mean()
print(dfg.sort_values(ascending=False).index[0])
```

```python
# sec 37
# 1. 각 행에서 판매수량과 단가를 이용하여 매출액을 계산하시오.
# 2. 요일을 평일과 주말로 구분하고, 매장별 평일과 주말 매출액 합계를 구하시오.
# 3. 매장별 평일과 주말 매출액 차이를 절대값으로 구하시오. 이후, 모든 매장 중 가장 큰 절댓값 차이를 찾으시오..
df = pd.read_csv('store_sales.csv')
df['매출'] = df['판매수량'] * df['단가']
weekend = ['토', '일']
df['주말'] = df['요일'].isin(weekend)
dfg = df.groupby(by=['매장코드','주말'])['매출'].sum().reset_index()
code2diff = dict()
for code in dfg.매장코드.unique():
    temp = dfg[dfg.매장코드==code]
    diff = temp.iloc[0]['매출'] - temp.iloc[1]['매출']
    code2diff[code] = abs(diff)
print(max(code2diff.values()))
```

```python
# sec 38
# 1. 각 region과 channel 조합별로 제품 A, B의 총 판매액을 계산하시오
# 2. 제품 A의 매출비율을 구하시오.
# 3. A 비율 중 최댓값을 찾아 반올림하여 소수 둘째자리까지 구하시오.
df = pd.read_csv('region_sales.csv')
# dfg = df.groupby(by=['Region', 'Channel', 'Product'])['Sales'].sum()
pdf = pd.pivot_table(df, index=['Region', 'Channel'], columns='Product', values='Sales', aggfunc='sum')
pdf['total'] = pdf['A'] + pdf['B']
pdf['A_ratio'] = pdf['A'] / pdf['total'] 
print(round(pdf['A_ratio'].max(),2))
```

```python
import pandas as pd
df = pd.DataFrame({
    '전': [6.1, 5.9, 6.3, 5.5, 6.0, 5.8, 6.2, 5.7, 6.1, 5.6],
    '후': [6.8, 6.5, 7.1, 6.0, 6.7, 6.3, 7.0, 6.4, 6.9, 6.2]
})
```

```python
from scipy import stats
t, p = stats.ttest_rel(df['전'],df['후'])
print(t,p)
```

```python
import pandas as pd
df = pd.DataFrame({
    'advertising': [10, 15, 20, 25, 30, 35, 40, 45, 50, 55],
    'visitors':    [100, 120, 140, 170, 180, 210, 230, 240, 270, 290],
    'sales':       [200, 250, 300, 380, 420, 480, 540, 580, 650, 710]
})
```

```python
import statsmodels.api as sm
X = df[['advertising', 'visitors']]
y = df['sales']

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(round(model.params, 3))
print(round(model.pvalues, 3))
print(round(model.rsquared, 3))
# model.params
```

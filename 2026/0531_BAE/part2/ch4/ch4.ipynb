```python
cols = ['Item_Fat_Content', 'Item_Type', 'Outlet_Identifier', 'Outlet_Size', 'Outlet_Location_Type', 'Outlet_Type']
```

```python
target = train.pop('Item_Outlet_Sales')
print(train.shape, test.shape)
df = pd.concat([train, test])
print(df.shape)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in cols:
    df[col] = le.fit_transform(df[col])

df.head()
```

```python
train = df.iloc[:len(train)].copy()
test  = df.iloc[len(train):].copy()
train.shape, test.shape
```

```python
train['Item_Weight'] = train['Item_Weight'].fillna(train['Item_Weight'].min())
train['Outlet_Size'] = train['Outlet_Size'].fillna(train['Outlet_Size'].mode()[0])

test['Item_Weight'] = test['Item_Weight'].fillna(test['Item_Weight'].min())
test['Outlet_Size'] = test['Outlet_Size'].fillna(test['Outlet_Size'].mode()[0])
```

```python
train = train.drop(columns = ['Item_Identifier'])
test  = test.drop(columns = ['Item_Identifier'])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(train, target, test_size=0.2, random_state=0)
X_train.shape, X_val.shape, y_train.shape, y_val.shape
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, root_mean_squared_error
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_val)

print(f'MSE: {mean_squared_error(y_pred, y_val):.2f}')
print(f'MAE: {mean_absolute_error(y_pred, y_val):.2f}')
print(f'R2: {r2_score(y_pred, y_val):.2f}')
print(f'RMSE: {root_mean_squared_error(y_pred, y_val):.2f}')
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=0)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_val)

print(f'MSE : {mean_squared_error(y_val,y_pred):.2f}')
print(f'MAE : {mean_absolute_error(y_val,y_pred):.2f}')
print(f'R2 : {r2_score(y_val,y_pred):.2f}')
print(f'RMSE : {root_mean_squared_error(y_val,y_pred):.2f}')
```

```python
import lightgbm as lgb
model = lgb.LGBMRegressor(random_state=0, verbose=1)
model.fit(X_train,y_train)
y_pred = model.predict(X_val)

print(f'MSE : {mean_squared_error(y_val,y_pred):.2f}')
print(f'MAE : {mean_absolute_error(y_val,y_pred):.2f}')
print(f'R2 : {r2_score(y_val,y_pred):.2f}')
print(f'RMSE : {root_mean_squared_error(y_val,y_pred):.2f}')
```

```python
pred = model.predict(test)
submit = pd.DataFrame({'pred':pred})
submit.to_csv('result.csv', index=False)
```
